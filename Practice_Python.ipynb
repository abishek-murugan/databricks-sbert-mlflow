1. Write a Python function that takes a list of integers and returns the second largest unique value.

Example:

Input:  [10, 5, 8, 10, 3, 8]

Output: 8

Requirements:

- Don't use sort() or sorted().
- Handle duplicate values correctly.
- Assume the list contains at least two unique values.

````python
def find_second_largest(lst):
  if len(lst) < 2:
    return None

  largest = float('-inf')
  second_largest = float('-inf')

  for num in lst:
    if num > largest:
      second_largest = largest
      largest = num
    elif num > second_largest and num != largest:
      second_largest = num

  if second_largest == float('-inf'):
    return None

  return second_largest

lst = [10, 5, 8, 10, 3, 8]
find_second_largest(lst)
````

2. Create a BankAccount class with:

- account_holder
- balance
- deposit(amount)
- withdraw(amount)

A withdrawal should only happen if the amount is positive and there is sufficient balance.

Write the class and demonstrate it with:

``` python
account = BankAccount("Abishek", 5000)
account.deposit(2000)
account.withdraw(1500)
print(account.balance)
```



````python
class BankAccount:
  def __init__(self, account_holder, balance):
    self.account_holder = account_holder
    self.balance = balance

  def withdraw(self,amount:int):
    if amount > self.balance:
      print("low balance")
    elif amount <= 0:
      print("enter a valid amount")
    else:
      self.balance = self.balance - amount
      print(f"Amount withdraw succesful, Available balance is Rs.{self.balance}")

  def deposit(self, amount:int):
      if amount <= 0:
        print("enter a valid amount")
      else:
        self.balance = self.balance + amount
        print(f"Amount deposit succesful, Available balance is Rs.{self.balance}")
````
